### Project Proposal

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Price Transparency as enforced by the Centers for Medicare and Medicaid Services (CMS) has made health pricing data publicy available for hospitals in the United States.<sup>1</sup>  Americans can make better informed decisions when determining healthcare costs.<sup>2</sup>  With Price Transparency, we can learn more about pricing schedules employed by hospitals. The aim of this project will be to examine whether pricing for hospital services is correlated with socio-economic factors, particularly examining hospitals and demographics within the corresponding ZIP Code. Our analysis will be limited to hospitals in California where socio-economic factors vary across the state and there are many hospitals.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Hospital pricing data are made available in readily-accessible, machine-readable files on hospital websites. We intend to use the American Hospital Directory (AHD) to collect a list of hospitals to be included in our study.<sup>3</sup>  The AHD reports hospital administrative data to help us collect pricing data. In conjunction, we intend to use data from the United States Census Bureau to obtain insight by ZIP Code.<sup>4</sup>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;To collect hospital data, we plan to employ search functionality made available on the AHD website. We narrow our search to hospitals in California where we will iterate through each available hospital to obtain the associated website, address, and other hospital level considerations. Once we have each hospital’s URL can we download the associated pricing data, typically available in a CSV. This may pose a challenge as each hospital may post their pricing data in varied manners. Thankfully, CMS has regulated there should be no barriers to obtain data (i.e. user logins, reCAPTCHA, user specific information). To collect socio-economic data, we plan to use the United States Census Bureau’s “Explore Census Data” functionality, particularly to identify income in the past twelve months by ZIP Code with other variables to be considered (e.g. race, population, age). This data was collected by the American Community Survey for 2019 (2020 currently not available).

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;We aim to employ web-scraping techniques to collect hospital pricing data. We will use data wrangling methods such as joining datasets (specifically results from each hospital and data from the Census Bureau) to then develop statistical models. We will use visualization to observe potential relations across geographic considerations.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Success of this project will be defined by reaching a point where we have collected data, joined data together to analyze at least three socio-economic factors, performed statistical analysis, and prepared visualizations to observe potential relationships between healthcare pricing and socio-economic factors.
